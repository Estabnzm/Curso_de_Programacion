```python
from IPython.display import display, clear_output
import ipywidgets as widgets

def display_menu():
    clear_output()  # limpia la pantalla
    print("Seleccionar una opción:")

    # Crear los botones
    button_ingresar = widgets.Button(description="Ingresar cliente")
    button_mostrar = widgets.Button(description="Mostrar Portafolio")
    button_calcular = widgets.Button(description="Calcular rendimiento")
    button_salir = widgets.Button(description="Salir")

    # Definir funciones de los botones
    def on_button_ingresar_clicked(b):
        clear_output()
        print("Ingresar cliente seleccionado")

    def on_button_mostrar_clicked(b):
        clear_output()
        print("Mostrar portafolio seleccionado")

    def on_button_calcular_clicked(b):
        clear_output()
        print("Calcular rendimiento seleccionado")

    def on_button_salir_clicked(b):
        clear_output()
        print("Saliendo del programa")

    # Conectar botones con funciones
    button_ingresar.on_click(on_button_ingresar_clicked)
    button_mostrar.on_click(on_button_mostrar_clicked)
    button_calcular.on_click(on_button_calcular_clicked)
    button_salir.on_click(on_button_salir_clicked)

    # Mostrar botones
    display(button_ingresar, button_mostrar, button_calcular, button_salir)

# Llamar al menú solo una vez
display_menu()
```

```python
# Diccionario para almacenar los datos de los cliente

portafolio = {}

# Función para ingresar un cliente
def ingresar_cliente():
    clear_output()  # Limpia la salida

    # Widgets para ingresar datos
    cliente = widgets.Text(description='Cliente', placeholder='Ingresa el Nombre')
    monto_invertido = widgets.FloatText(description='Monto Invertido', value=0.0)
    rendimiento_anual = widgets.FloatText(description='Rendimiento Anual (%)', value=0.0)

    # Botón para guardar cliente
    button = widgets.Button(description="Guardar")
    output = widgets.Output()

    def on_button_clicked(b):
        # Guardar los datos en el diccionario
        if cliente.value:
            portafolio[cliente.value] = {
                "Monto Invertido": monto_invertido.value,
                "Rendimiento Anual": rendimiento_anual.value
            }
            with output:
                clear_output()
                print(f"Cliente {cliente.value} guardado en el portafolio.")
        else:
            with output:
                clear_output()
                print("Por favor, ingresa un nombre de cliente.")

    button.on_click(on_button_clicked)

    display(cliente, monto_invertido, rendimiento_anual, button, output)

# Ejecutar la función para ingresar cliente
ingresar_cliente()









```

El valor presente neto(vpn): El valor presente neto (VNP) se calcula utilizando una tasa de descuento fija. Para cada proyecto, sumaremos el valor presente de todos los flujos de caja.
# El valor valor persente(VPN)
es una medida financiera que nos indica valor actual de los flujos de caja futuros de un proyecto, descontando una tasa especifica. Esta tasa, llamasa Tasa de descuento, represnta del costo de la oportunidad de capital o rendimiento minimo que el inversor espera obtener.

# calculo de la tasa interna de Retorno(tir)
Definimos una función para calcular TIR para cada proyectonusando Numpy Financial

# La Tasa Interna de Retorno (Tir)
es la tasa de descuentos que hace  que el VPN sea igual a cero, es decir es la tasa que equilibra los flujos futuros y la inversión inicial, lo que significa que el proyecto pierde valor neto.


```python
# TRM

for col in df.columns[1:]:
    df[col] = df[col].replace('[^\d\.-]', '', regex=True).astype(float)

# Tasa de descuento fija
r = 0.10

# Función para calcular VPN usando la fórmula tradicional
def calcular_vpn(row, tasa):
    vpn = 0
    num_flujos = len([col for col in df.columns if col.startswith('Año_')])
    for t in range(0, num_flujos + 1):  # Incluye t = 0
        if t == 0:
            flujo = row['Inversion_Inicial']
        else:
            col_name = f'Año_{t}'
            flujo = row[col_name] if col_name in row else 0
        valor_presente = flujo / ((1 + tasa) ** t)
        vpn += valor_presente
    return vpn

# Aplicar cálculo a cada fila
df['VPN'] = df.apply(lambda row: calcular_vpn(row, r), axis=1)
print(df[['Proyecto', 'VPN']])
```

```python
import numpy_financial as npf

# Función para obtener lista de flujos de caja por proyecto
def obtener_flujos(row):
    flujos = [row['Inversion_Inicial']]  # Año 0
    for t in range(1, 11):  # Años 1 a 10
        col_name = f'Año_{t}'
        flujos.append(row[col_name] if col_name in row else 0)
    return flujos

# Función para calcular TIR
def calcular_tir(row):
    flujos = obtener_flujos(row)
    return npf.irr(flujos)

# Aplicar TIR a cada fila del DataFrame
df['TIR'] = df.apply(calcular_tir, axis=1)

# Mostrar resultados
print(df[['Proyecto', 'TIR']])
```

```python
# 4. GRAFICA CON MATPLOTLIB DE VPN
import matplotlib.pyplot as plt

df_sorted = df.sort_values(by='VPN', ascending=False)

plt.figure(figsize=(14, 8)) # Ajustar tamaño para mejor legibilidad

# Crear el gráfico de barras
bars = plt.bar(df_sorted['Proyecto'], df_sorted['VPN'], color='skyblue')

# Añadir títulos y etiquetas
plt.title('Valor Presente Neto (VPN) por Proyecto', fontsize=16)
plt.xlabel('Proyecto', fontsize=12)
plt.ylabel('VPN (en unidades monetarias)', fontsize=12)
```
