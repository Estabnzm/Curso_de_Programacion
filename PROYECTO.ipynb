```python
import pandas as pd

url = "https://drive.google.com/uc?export=download&id=1SzEHZlD1SomHSKFBaWlekyTHgJSBq3Gz"
df = pd.read_excel(url)

df.head()
```

#**EXPLORACIÓN BASE DE DATOS**

```python
# Información dataframe
df.head()
```

#**Análisis estadístico base de datos**

```python
df.describe(include='all')
```

#**CÁLCULO RENTABILIDAD MENSUAL x AÑO**


```python
# Rentabilidad mensual usando la fórmula: (mes actual - mes anterior) / mes anterior

rentabilidad = (df - df.shift(1)) / df.shift(1)
print(rentabilidad)
```

#**NUEVO ARCHIVO DE RENTABILIDADES**

```python
rentabilidad
```

#**RENTABILIDAD PROMEDIO MENSUAL PARA CADA ACCIÓN**

```python
# Calcular rentabilidad promedio por acción (columna)
promedio_por_accion = rentabilidad.mean(axis=0, skipna=True)

# Mostrar resultado
print(promedio_por_accion)
```

#**NUEVO ARCHIVO DE PROMEDIO MESUAL**

```python
promedio_por_accion.to_excel('promedio_por_accion', header=['Promedio'])
```

#**RENTABILIDAD MÍNIMA DE LAS ACCIONES**

```python
# Rentabilidad mínima de las acciones

rentabilidad_minima = rentabilidad.min().min()

print(f"Rentabilidad mínima observada: {rentabilidad_minima:.4f}")
```

#**ACCIÓN CON LA RENATABILIDAD MÁS BAJA**

```python
# Ubicar la posición (fecha y acción) de la rentabilidad mínima

min_location = rentabilidad.stack().idxmin()
min_value = rentabilidad.min().min()

print(f"La rentabilidad mínima observada fue de {min_value:.4f}, en {min_location[0].date()} para la acción {min_location[1]}")
```

#**RENTABILIDAD MÁXIMA DE LAS ACCIONES**

```python
# Rentabilidad máxima de las acciones

rentabilidad_max = rentabilidad.max().max()

print(f"Rentabilidad maxima observada: {rentabilidad_max:.4f}")
```

#**ACCIÓN CON MAYOR RENTABILIDAD**

```python
# Ubicar la posición (fecha y acción) de la rentabilidad máxima

max_location = rentabilidad.stack().idxmax()
max_value = rentabilidad.max().max()

print(f"La rentabilidad máxima observada fue de {max_value:.4f}, en {max_location[0].date()} para la acción {max_location[1]}")
```

#**RENTABILIDAD ACUMULADA TOTAL**

```python

# Calcular rentabilidad acumulada

precio_inicial = df.iloc[0]  # Primer precio de cada acción
precio_final = df.iloc[-1]   # Último precio de cada acción

# Aplicar la fórmula de rentabilidad acumulada: ((precio final / precio inicial) - 1) * 100

rentabilidad_acumulada = ((precio_final / precio_inicial) - 1) * 100

# Mostrar rentabilidad acumulada por acción

print(rentabilidad_acumulada)
```
